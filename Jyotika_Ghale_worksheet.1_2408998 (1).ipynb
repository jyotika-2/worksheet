**• For the following arrays:
A = np.array([[3,4],[7,8]]) and B = np.array([[5,3],[2,1]]);**

**Prove following with Numpy:**
**1. Prove A.A−1 = I.**
**2. Prove AB ̸= BA.**
**3. Prove (AB)**

**T = BTAT
.**

```python
import numpy as np

A = np.array([[3, 4], [7, 8]])
B = np.array([[5, 3], [2, 1]])

# 1. Prove A * A^-1 = I
A_inv = np.linalg.inv(A)
result = np.dot(A, A_inv)
identity_matrix = np.eye(2)
print("A * A^-1:")
print(result)
print("Is A * A^-1 approximately I?", np.allclose(result, identity_matrix))

# 2. Prove AB != BA
AB = np.dot(A, B)
BA = np.dot(B, A)
print("AB:")
print(AB)
print("BA:")
print(BA)
print("Is AB equal to BA?", np.array_equal(AB, BA))

# 3. Prove (AB)^T = B^T A^T
AB_transpose = np.transpose(np.dot(A, B))
BT_AT = np.dot(np.transpose(B), np.transpose(A))
print("(AB)^T:")
print(AB_transpose)
print("B^T A^T:")
print(BT_AT)
print("Is (AB)^T equal to B^T A^T?", np.array_equal(AB_transpose, BT_AT))
```

**• Solve the following system of Linear equation using Inverse Methods.**

**x − y + 2z = −3**

**3x + y − z = 9**

**• Now: solve the above equation using np.linalg.inv function.{Explore more about ”linalg” function
of Numpy}**

```python
import numpy as np

A = np.array([[2, -3, 1], [1, -1, 2], [3, 1, -1]])
B = np.array([-1, -3, 9])

A_inv = np.linalg.inv(A)
X_inv = np.dot(A_inv, B)

X_solve = np.linalg.solve(A, B)

det_A = np.linalg.det(A)

eigenvalues, eigenvectors = np.linalg.eig(A)

norm_B = np.linalg.norm(B)

print("Solution using np.linalg.inv():")
print(X_inv)

print("\nSolution using np.linalg.solve():")
print(X_solve)

print("\nDeterminant of A:")
print(det_A)

print("\nEigenvalues of A:")
print(eigenvalues)

print("\nEigenvectors of A:")
print(eigenvectors)

print("\nNorm of B:")
print(norm_B)
```
