**To - Do:4**
We will define a function that:
1. Loads the data and splits it into training and test sets.
2. Prepares the feature matrix (X) and target vector (Y).
3. Defines the weight matrix (W) and initializes the learning rate and number of iterations.
4. Calls the gradient descent function to learn the parameters.
5. Evaluates the model using RMSE and R2

```python
def train_and_evaluate_model(data, learning_rate=0.01, num_iterations=1000):
    # Prepare the data
    X = data.drop('Math', axis=1).values
    Y = data['Math'].values.reshape(-1, 1)  # Reshape Y to a column vector

    # Normalize features (important for gradient descent)
    X_mean = np.mean(X, axis=0)
    X_std = np.std(X, axis=0)
    X = (X - X_mean) / X_std

    # Split the data
    split_ratio = 0.8
    split_index = int(len(X) * split_ratio)
    X_train, X_test = X[:split_index], X[split_index:]
    Y_train, Y_test = Y[:split_index], Y[split_index:]

    # Initialize weights
    num_features = X_train.shape[1]
    W = np.random.rand(num_features, 1)  # Initialize weights randomly


    # Gradient descent function (implementation not shown but assumed to exist)
    def gradient_descent(X, Y, W, learning_rate, num_iterations):
      for _ in range(num_iterations):
          # Calculate predictions
          Y_pred = np.dot(X, W)

          # Calculate the error
          error = Y_pred - Y

          # Calculate the gradient
          gradient = np.dot(X.T, error) / len(X)

          # Update weights
          W = W - learning_rate * gradient

      return W

    W = gradient_descent(X_train, Y_train,W, learning_rate, num_iterations)

    # Make predictions on test data
    Y_pred_test = np.dot(X_test, W)


    # Evaluate the model
    def rmse(y_true, y_predicted):
        return np.sqrt(np.mean((y_true - y_predicted)**2))

    def r2_score(y_true, y_predicted):
        mean_y_true = np.mean(y_true)
        ss_tot = np.sum((y_true - mean_y_true)**2)
        ss_res = np.sum((y_true - y_predicted)**2)
        r2 = 1 - (ss_res / ss_tot)
        return r2

    rmse_val = rmse(Y_test,Y_pred_test)
    r2 = r2_score(Y_test,Y_pred_test)

    print(f"RMSE: {rmse_val}")
    print(f"R^2: {r2}")

    return W # Return the trained weights

trained_weights = train_and_evaluate_model(data)
```
